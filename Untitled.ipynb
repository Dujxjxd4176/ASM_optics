```python
from PIL import Image , ImageFilter
import numpy as np
import time
img = Image.open("./aperture/product/5fz.bmp")
#first resize the thng 
pixelsize = 1
pensize  = 1.5
stepsize =1
coord = np.array(img.size)
coord *= pixelsize
print(coord)
coloumOffset = round(pensize/stepsize)
print(coloumOffset)
resizeimg = img.resize(np.rint(coord/stepsize).astype(int))
resizeimg.show()
print(resizeimg.size)


```

```python
alterimg = np.array(resizeimg)

print(alterimg)
print (sum(sum(alterimg)))
im = Image.fromarray(alterimg)
im.show()
```

```python
# state machine in resizing objects 
# first it goes through all the parts if it hit the edge of the object 
# it would go back and shrink it according to the fomrula 
start = time.time()
for vy, j in  enumerate(alterimg):
    inobj = False
    #print(time.time()-start)
    for vx, i in enumerate(j):
        if((inobj == True and i == True ) ):
            for k in range (coloumOffset):
                if (vx -k-1) < 0:
                    break
                if(alterimg[vy][vx-k-1] == True ):
                    break
                alterimg[vy][vx-k-1] = True
            inobj = False
        elif(vx == j.size -1  and i == False):
            for k in range (coloumOffset):
                if (vx -k) < 0:
                    break
                if(alterimg[vy][vx-k] == True ):
                    break
                alterimg[vy][vx-k] = True
            inobj = False
        elif(i == False):
            inobj = True
print(alterimg)
im = Image.fromarray(alterimg)
im.show()  
print (sum(sum(alterimg)))

print(time.time()-start)
```
